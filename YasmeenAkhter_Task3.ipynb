### Project Objective
Demonstrate professional-level data cleaning skills by taking a messy dataset, producing a data quality report, handling missing values, removing duplicates, normalising formatting, detecting outliers, correcting data types, and producing a "Before vs After" summary.

### Step 1: Load Dataset & Data Quality Report
We load the dataset and analyze its quality by checking null values, duplicate rows, data types, and unique values to spot anomalies.

```python
import pandas as pd
import numpy as np

# Dataset load karein
df = pd.read_csv('retail_sales_dataset.csv')

print("--- DATA QUALITY REPORT ---")
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
print("\n--- Null Values per Column ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows Count ---")
print(df.duplicated().sum())

print("\n--- Data Types ---")
print(df.dtypes)
display(df.head())
```

### Step 2: Missing Data Handling & Duplicate Removal
We check for duplicates and handle missing values using appropriate imputation strategies (such as mode, median, or deletion) with proper justifications.

```python
# Duplicate removal (agar koi hon)
initial_rows = df.shape[0]
df = df.drop_duplicates()
removed_duplicates = initial_rows - df.shape[0]
print(f"Duplicate rows removed: {removed_duplicates}")

# Missing values check & handling strategy
# (Agar dataset me missing values hon toh median/mode imputation use karein)
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype == 'object':
            df[col].fillna(df[col].mode()[0], inplace=True)
        else:
            df[col].fillna(df[col].median(), inplace=True)

print("Missing values after treatment:\n", df.isnull().sum())
```

### Step 3: Standardisation and Data Type Correction
We normalise formatting (e.g., text capitalization, date formats to datetime) and ensure all columns have the correct data types.

```python
# Date column ko datetime me convert karna
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])

# Text columns ki formatting standardise karna (e.g., Gender ya Product Category)
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].str.capitalize()

# IDs ko string me convert karna
if 'Customer ID' in df.columns:
    df['Customer ID'] = df['Customer ID'].astype(str)

print("Data types corrected successfully!")
print(df.dtypes)
```

### Step 4: Outlier Detection (IQR Method)
We use the Interquartile Range (IQR) method to identify outliers in numeric columns (like Total Amount or Price per Unit) and decide whether to cap or retain them.

```python
numeric_cols = ['Quantity', 'Price per Unit', 'Total Amount']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    print(f"Outliers detected in {col}: {outliers_count}")
```

### Step 5: Before vs. After Summary & Export
We create a comparison summary table and export the cleaned dataset to a new CSV file.

```python
# Before vs After Summary DataFrame
summary_data = {
    "Metric": ["Row Count", "Null Count (Total)", "Duplicate Count", "Date Dtype"],
    "Before Cleaning": [1000, 0, 0, "object"],
    "After Cleaning": [df.shape[0], df.isnull().sum().sum(), df.duplicated().sum(), str(df['Date'].dtype)]
}

summary_df = pd.DataFrame(summary_data)
display(summary_df)

# Cleaned dataset ko nayi CSV file me save karna
df.to_csv('cleaned_retail_sales_dataset.csv', index=False)
print("\nCleaned dataset successfully saved as 'cleaned_retail_sales_dataset.csv'!")
```
